```python
spark.conf.set("spark.sql.adaptive.enabled", "false")
```

```python
from pyspark.sql.functions import *
```

```python
# Big DatFrame: treat it as big data frame
df_transactions = spark.createDataFrame([
    (1,"US",100),(2,"IN",200),(3,"UK",150),(4,"IRE",80),(5,"SA",96),(6,"AUS",220)
],["id","country_code","amount"])
# Small DataFrame
df_countries = spark.createDataFrame([("US","United States"),("IN","India"),("UK","United Kingdom")],["country_code","country_name"])
```

```python
df_transactions.display()
df_countries.display()
```

```python
df_join = df_transactions.join(df_countries, df_transactions['country_code'] == df_countries['country_code'],'inner')
```

```python
df_join.display()
```

#### Broadcast Join

```python
df_broad_join = df_transactions.join(broadcast(df_countries), df_transactions['country_code'] == df_countries['country_code'],'inner')
```

```python
df_broad_join.display()
```
