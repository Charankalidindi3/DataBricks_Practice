```python
from pyspark.sql.functions import *
```

```python
# Dataframe
df = spark.createDataFrame([
    ("1001",),
    ("1002",),
    ("1003",),
    ("1004",)
],["product_id"])

#lookup dicionary (small)

product_dict ={
    "1001": "iPhone",
    "1002":"Samsung",
    "1003": "Pixel"
}
```

```python
# Broadcasting the dictionary varioble
broad_var = spark.sparkContext.broadcast(product_dict)
```

```python
print(broad_var.value)
```

```python
# our  function
def mymap(x):
    return broad_var.value.get(x)
```

```python
mymap_udf =udf(mymap)
```

```python
df_with_names = df.withColumn("product_name",mymap_udf("product_id"))
```
